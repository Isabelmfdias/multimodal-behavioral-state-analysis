```python
from pathlib import Path
import pandas as pd

# =========================
# PROJECT PATHS
# =========================

PROJECT_ROOT = Path.cwd().parents[1]
DATA_FEB_DIR = PROJECT_ROOT / "Dataset_feb"
SAMSUNG_DIR = DATA_FEB_DIR / "Dataset_samsung"

input_file = SAMSUNG_DIR / "Geo_file.csv"
output_file = SAMSUNG_DIR / "Geo_file_Cleaned.csv"

print("Input file found:", input_file.exists())


# =========================
# LOAD DATA
# =========================

if input_file.exists():
    df = pd.read_csv(input_file)

    # =========================
    # REMOVE IMPLAUSIBLE ALTITUDE VALUES
    # =========================

    df_cleaned = df[df["altitude"] <= 1995].copy()

    # =========================
    # SAVE CLEANED DATA
    # =========================

    df_cleaned.to_csv(output_file, index=False)

    # =========================
    # SUMMARY
    # =========================

    print("Cleaning completed successfully!")
    print(f"Original rows: {len(df)}")
    print(f"Rows after cleaning: {len(df_cleaned)}")
    print(
        f"Removed altitude points: "
        f"{len(df) - len(df_cleaned)}"
    )
    print(f"Cleaned file saved to: {output_file}")

else:
    print(f"Input file not found: {input_file}")
```
