```python
from pathlib import Path
import pandas as pd

# =========================
# 1. PROJECT PATHS
# =========================

PROJECT_ROOT = Path.cwd().parents[1]
DATA_FEB_DIR = PROJECT_ROOT / "Dataset_feb"

PARTICIPANTS_FILE = DATA_FEB_DIR / "Participants_file.csv"
OUTPUT_FILE = DATA_FEB_DIR / "participants-file-com-bmi.xlsx"

print("Input file:", PARTICIPANTS_FILE)
print("Input file exists:", PARTICIPANTS_FILE.exists())

# Load participant data
df = pd.read_csv(PARTICIPANTS_FILE)


# =========================
# 2. BMI CALCULATION
# =========================

# Convert height from centimeters to meters
df["height_m"] = df["height"] / 100

# Calculate BMI
df["bmi"] = df["weight"] / (df["height_m"] ** 2)

# Round BMI to two decimal places
df["bmi"] = df["bmi"].round(2)


# =========================
# 3. BMI CLASSIFICATION
# =========================

def classify_bmi(bmi):
    if pd.isna(bmi):
        return None
    elif bmi < 18.5:
        return "Baixo peso"
    elif bmi < 25:
        return "Peso normal"
    elif bmi < 30:
        return "Excesso de peso"
    elif bmi < 35:
        return "Obesidade Classe 1"
    elif bmi < 40:
        return "Obesidade Classe 2"
    else:
        return "Obesidade Classe 3"


def disease_risk(bmi):
    if pd.isna(bmi):
        return None
    elif bmi < 18.5:
        return "Baixo"
    elif bmi < 25:
        return "Médio"
    elif bmi < 30:
        return "Aumentado"
    elif bmi < 35:
        return "Alto"
    elif bmi < 40:
        return "Muito Alto"
    else:
        return "Extremo"


df["bmi_class"] = df["bmi"].apply(classify_bmi)
df["disease_risk"] = df["bmi"].apply(disease_risk)


# =========================
# 4. REMOVE TEMPORARY COLUMN
# =========================

df.drop(columns=["height_m"], inplace=True)


# =========================
# 5. SAVE PROCESSED DATA
# =========================

df.to_excel(OUTPUT_FILE, index=False)

print("File created successfully!")
print("Saved to:", OUTPUT_FILE)

# Preview selected columns
print("\nPreview:")
print(
    df[
        [
            "user_id",
            "weight",
            "height",
            "bmi",
            "bmi_class",
            "disease_risk"
        ]
    ].head()
)
```
