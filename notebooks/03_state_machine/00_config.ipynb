This notebook defines the project paths, temporal configuration parameters, and profile-specific thresholds used throughout the finite state machine analysis

```python
from pathlib import Path

# =========================
# PROJECT PATHS
# =========================

PROJECT_ROOT = Path.cwd().parents[1]

DATA_FEB_DIR = PROJECT_ROOT / "Dataset_feb"
SAMSUNG_DIR = DATA_FEB_DIR / "Dataset_samsung"
ACTIGRAPH_DIR = DATA_FEB_DIR / "Dataset_actigraph"

# State machine outputs
OUTPUT_DIR = PROJECT_ROOT / "outputs" / "state_machine"
RESULTS_DIR = OUTPUT_DIR / "results"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)


# =========================
# INPUT FILES
# =========================

PARTICIPANTS_FILE = (
    DATA_FEB_DIR
    / "participants-file-com-bmi.xlsx"
)

ACTIVITY_FILE = (
    SAMSUNG_DIR
    / "Activity_file.csv"
)

DAILY_ACTIVITY_FILE = (
    SAMSUNG_DIR
    / "Daily_activity_file.csv"
)

MOOD_FILE = (
    SAMSUNG_DIR
    / "Mood_file.csv"
)

ACTIGRAPH_FILE = (
    ACTIGRAPH_DIR
    / "Scoring all data_Freedson2011.xlsx"
)

ACTIGRAPH_DAILY_ACTIVITY_FILE = (
    ACTIGRAPH_DIR
    / "Daily_activity_file.csv"
)


# =========================
# FILE CHECK
# =========================

print("Project root:", PROJECT_ROOT)
print("Samsung data folder:", SAMSUNG_DIR)
print("Actigraph data folder:", ACTIGRAPH_DIR)

print("\nInput files:")
print("Participants file found:", PARTICIPANTS_FILE.exists())
print("Activity file found:", ACTIVITY_FILE.exists())
print("Daily activity file found:", DAILY_ACTIVITY_FILE.exists())
print("Mood file found:", MOOD_FILE.exists())
print("Actigraph file found:", ACTIGRAPH_FILE.exists())
print(
    "Actigraph daily activity file found:",
    ACTIGRAPH_DAILY_ACTIVITY_FILE.exists()
)

print("\nOutput folder:", OUTPUT_DIR)
print("Results folder:", RESULTS_DIR)


# =========================
# FSM TEMPORAL PARAMETERS
# =========================

MIN_DAYS_FOR_FSM = 7


def choose_user_config(n_days):

    if n_days >= 14:
        return {
            "onboarding_days": 7,
            "window_days": 7,
            "max_transition_days": 14
        }

    elif n_days >= 10:
        return {
            "onboarding_days": 5,
            "window_days": 5,
            "max_transition_days": 10
        }

    elif n_days >= 7:
        return {
            "onboarding_days": 3,
            "window_days": 3,
            "max_transition_days": 7
        }

    else:
        return None


# =========================
# PROFILE-BASED THRESHOLDS
# =========================

PROFILE_THRESHOLDS = {

    "healthy": {
        "steps_transition": 5000,
        "mvpa_transition": 20,
        "mvpa_low": 75,
        "mvpa_active": 150,
        "cv_regular": 45,
        "mad_k": 2.5
    },

    "overweight_obese": {
        "steps_transition": 4000,
        "mvpa_transition": 15,
        "mvpa_low": 60,
        "mvpa_active": 120,
        "cv_regular": 45,
        "mad_k": 0.75
    }
}


def get_profile_thresholds(bmi_profile):
    """Return the FSM thresholds associated with a BMI profile."""

    bmi_profile = str(bmi_profile).lower()

    if bmi_profile in PROFILE_THRESHOLDS:
        return PROFILE_THRESHOLDS[bmi_profile]

    return PROFILE_THRESHOLDS["healthy"]
```
